# Superstore EDA 与可视化示例
使用 pandas 进行数据清洗，Pyecharts 与 Matplotlib 做可视化。

```python
import pandas as pd
from pyecharts.charts import Line
from pyecharts import options as opts

df = pd.read_csv('../data/superstore_sample.csv')
df['Order Date'] = pd.to_datetime(df['Order Date'])
df.head()
```

```python
# 月度销售折线图示例
ts = df.groupby(pd.Grouper(key='Order Date', freq='M'))['Sales'].sum().reset_index()
x = ts['Order Date'].dt.strftime('%Y-%m').tolist()
y = ts['Sales'].round(2).tolist()

line = (
    Line()
    .add_xaxis(x)
    .add_yaxis('Monthly Sales', y)
    .set_global_opts(title_opts=opts.TitleOpts(title='Monthly Sales'))
)
line.render('../reports/notebook_monthly_sales.html')
print('rendered to reports/notebook_monthly_sales.html')
```
